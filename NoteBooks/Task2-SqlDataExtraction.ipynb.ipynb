## Advanced SQL

```python
# 1. Subquery
query = """
SELECT
    "Product Name",
    Sales
FROM sales
WHERE Sales > (
    SELECT AVG(Sales)
    FROM sales
)
ORDER BY Sales DESC;
"""

pd.read_sql(query, conn)
```

```python
# CTE=Common Table Expression:
query = """
WITH category_sales AS (
    SELECT
        Category,
        SUM(Sales) AS Total_Sales
    FROM sales
    GROUP BY Category
)
SELECT *
FROM category_sales
ORDER BY Total_Sales DESC;
"""

pd.read_sql(query, conn)
```

```python
# Window Function
query = """
SELECT
    "Product Name",
    Sales,
    RANK() OVER (
        ORDER BY Sales DESC
    ) AS Sales_Rank
FROM sales
LIMIT 20;
"""

pd.read_sql(query, conn)
```

```python
# LAG
query = """
SELECT
    "Order Date",
    Sales,
    LAG(Sales) OVER (
        ORDER BY "Order Date"
    ) AS Previous_Sales
FROM sales
LIMIT 20;
"""

pd.read_sql(query, conn)
```

```python
# Create a SQL View
query = """
CREATE VIEW IF NOT EXISTS category_sales AS
SELECT
    Category,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY Category;
"""

conn.execute(query)
conn.commit()
```

```python
query = """
SELECT *
FROM category_sales
ORDER BY Total_Sales DESC;
"""

pd.read_sql(query, conn)
```

## Creating a Reusable Database Utility

```python
# Creating a reusable database utility script
import sqlite3
import pandas as pd

def create_connection(database):
    return sqlite3.connect(database)

def load_csv_to_database(csv_file, database, table_name):
    conn = sqlite3.connect(database)

    df = pd.read_csv(csv_file)

    df.to_sql(
        table_name,
        conn,
        if_exists="replace",
        index=False
    )

    conn.close()

def run_query(database, query):
    conn = sqlite3.connect(database)

    result = pd.read_sql(query, conn)

    conn.close()

    return result
```

## Subquery

```python
query = """
SELECT
    "Product Name",
    SUM(Sales) AS Total_Sales
FROM sales
GROUP BY "Product Name"
HAVING SUM(Sales) >
(
    SELECT AVG(Product_Sales)
    FROM
    (
        SELECT
            "Product Name",
            SUM(Sales) AS Product_Sales
        FROM sales
        GROUP BY "Product Name"
    )
)
ORDER BY Total_Sales DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# CTE=Common Table Expression:
query = """
WITH product_sales AS
(
    SELECT
        "Product Name",
        SUM(Sales) AS Total_Sales
    FROM sales
    GROUP BY "Product Name"
)

SELECT *
FROM product_sales
ORDER BY Total_Sales DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
# ROW_NUMBER()
query = """
SELECT
    "Product Name",
    SUM(Sales) AS Total_Sales,
    ROW_NUMBER() OVER (
        ORDER BY SUM(Sales) DESC
    ) AS Product_Number
FROM sales
GROUP BY "Product Name";
"""

pd.read_sql_query(query, conn)
```

```python
# RANK()
query = """
SELECT
    Category,
    SUM(Sales) AS Total_Sales,
    RANK() OVER (
        ORDER BY SUM(Sales) DESC
    ) AS Sales_Rank
FROM sales
GROUP BY Category;
"""

pd.read_sql_query(query, conn)
```

```python
# LAG()query = """
query = """
WITH monthly_sales AS (
    SELECT
        strftime('%Y-%m', "Order Date") AS Month,
        SUM(Sales) AS Total_Sales
    FROM sales
    GROUP BY strftime('%Y-%m', "Order Date")
)

SELECT
    Month,
    Total_Sales,
    LAG(Total_Sales) OVER (
        ORDER BY Month
    ) AS Previous_Month_Sales
FROM monthly_sales
ORDER BY Month;
"""

pd.read_sql_query(query, conn)
```

```python
# LEAD()
query = """
WITH monthly_sales AS
(
    SELECT
        strftime('%Y-%m', "Order Date") AS Month,
        SUM(Sales) AS Total_Sales
    FROM sales
    GROUP BY strftime('%Y-%m', "Order Date")
)

SELECT
    Month,
    Total_Sales,
    LEAD(Total_Sales) OVER (
        ORDER BY Month
    ) AS Next_Month_Sales
FROM monthly_sales
ORDER BY Month;
"""

pd.read_sql_query(query, conn)
```
