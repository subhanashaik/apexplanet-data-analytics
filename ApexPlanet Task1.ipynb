## 3.Data Cleaning

```python
#  Checking Missing Values
df.isnull().sum()
```

```python
# for numerical values
df["Sales"]=df["Sales"].fillna(df["Sales"].median())
# for categorical columns
df["Category"]= df["Category"].fillna(df["Category"].mode()[0])
df = df.dropna()
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.dtypes
```

```python
print(df.columns)
```

```python
df.columns = df.columns.str.strip()
```

```python
print(df.columns.tolist())
```

```python
for col in df.columns:
    print(repr(col))
```

```python
df.head(5)
```

```python
df["Sales"] = pd.to_numeric(df["Sales"], errors="coerce")
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["Discount"] = pd.to_numeric(df["Discount"], errors="coerce")
df["Profit"] = pd.to_numeric(df["Profit"], errors="coerce")
```

```python
df["Ship Mode"] = df["Ship Mode"].astype("string")
df["Segment"] = df["Segment"].astype("string")
df["Country"] = df["Country"].astype("string")
df["City"] = df["City"].astype("string")
df["State"] = df["State"].astype("string")
df["Region"] = df["Region"].astype("string")
df["Category"] = df["Category"].astype("string")
df["Sub-Category"] = df["Sub-Category"].astype("string")
```

```python
# Handle Outliers using IQR
Q1 = df["Sales"].quantile(0.25)
Q3 = df["Sales"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print("Lower:", lower_bound)
print("Upper:", upper_bound)
```

```python
outliers = df[(df["Sales"] < lower_bound) | (df["Sales"] > upper_bound)]
print("Number of outliers:",len(outliers))
```

```python
print(outliers[["Sales"]].head(20))
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], format="mixed")
```

```python
if df.isnull().sum().sum() == 0 and df.duplicated().sum() == 0:
    print(" NO Missing Values OR  duplicates rows found")
else:
    print("Data needs cleaning")
```

## 4.Exploratary Data Analysis 

## Histogram

```python
# Create Histogram
plt.figure(figsize=(8,5))
plt.hist(df["Sales"], bins=30 , color="Red",edgecolor="Black",label="Sales")
plt.title("Distribution of Sales",fontsize=18, fontweight="bold",color="DarkBlue")
plt.xlabel("Sales", fontsize=12,fontweight="bold")
plt.ylabel("Frequency",fontsize=12,fontweight="bold")
plt.legend()
plt.show()
```

## Box Plot

```python
#Box Plot
plt.figure(figsize=(8, 5))
sns.boxplot( y=df["Sales"], color="skyblue")
plt.title("Sales Distribution and Outliers",fontsize=18,fontweight="bold",color="darkblue")
plt.ylabel("Sales",fontsize=12,fontweight="bold")
plt.show()
```

## Bar charts

```python
# Bar Chart---Sales by Region
region_sales = df.groupby("Region")["Sales"].sum()
plt.figure(figsize=(8, 5))
plt.bar(region_sales.index,region_sales.values,color="orange",edgecolor="black",label="Sales")
plt.title("Sales by Region",fontsize=18,fontweight="bold",color="darkblue")
plt.xlabel("Region",fontsize=12,fontweight="bold")
plt.ylabel("Total Sales",fontsize=12,fontweight="bold")
plt.legend()
plt.show()
```

```python
#Bar Chart — Sales by Segment
segment_sales = df.groupby("Segment")["Sales"].sum()
plt.figure(figsize=(8, 5))
plt.bar(segment_sales.index,segment_sales.values,color="lightgreen",edgecolor="black", label="Sales")
plt.title("Sales by Segment",fontsize=18,fontweight="bold",color="darkblue")
plt.xlabel("Segment",fontsize=12,fontweight="bold")
plt.ylabel("Total Sales",fontsize=12,fontweight="bold")
plt.legend()
plt.show()
```

```python
#Bar Chart — Sales by Category
category_sales = df.groupby("Category")["Sales"].sum()
plt.figure(figsize=(8, 5))
plt.bar(category_sales.index,category_sales.values,color="plum",edgecolor="black", label="Sales")
plt.title( "Sales by Category", fontsize=18, fontweight="bold",color="darkblue")
plt.xlabel("Category",fontsize=12, fontweight="bold")
plt.ylabel("Total Sales",fontsize=12,fontweight="bold")
plt.legend()
plt.show()
```

```python
# Bar Chart — Profit by Category
category_profit = df.groupby("Category")["Profit"].sum()
plt.figure(figsize=(8, 5))
plt.bar(category_profit.index,category_profit.values,color="lightcoral", edgecolor="black", label="Profit")
plt.title("Profit by Category", fontsize=18,fontweight="bold", color="darkblue")
plt.xlabel("Category",fontsize=12,fontweight="bold")
plt.ylabel("Total Profit",fontsize=12, fontweight="bold")
plt.legend()
plt.show()
```

```python
#Correlation Heatmap
numeric_df = df.select_dtypes(include=np.number)
correlation = numeric_df.corr()
plt.figure(figsize=(10, 7))
sns.heatmap( correlation, annot=True,cmap="coolwarm",linewidths=0.5)
plt.title( "Correlation Heatmap",fontsize=18, fontweight="bold",color="darkblue")
plt.legend
plt.show()
```

## Line Charts

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], format="%m/%d/%Y")
monthly_sales = df.groupby(df["Order Date"].dt.month)["Sales"].sum()
monthly_sales.plot(kind="line", figsize=(10,5))
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
monthly_profit = df.groupby(df["Order Date"].dt.month)["Profit"].sum()

monthly_profit.plot(kind="line", figsize=(10,5))

plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

category_sales.plot(kind="line", figsize=(8,5), marker="o")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
order_date = pd.to_datetime(df["Order Date"], format="mixed")

monthly_order_sales = df.copy()
monthly_order_sales["Order Date"] = order_date
monthly_order_sales = monthly_order_sales.groupby(
    monthly_order_sales["Order Date"].dt.to_period("M")
)["Sales"].sum()

monthly_order_sales.plot(kind="line", figsize=(10,5))

plt.title("Monthly Sales by Order Date")
plt.xlabel("Order Month")
plt.ylabel("Total Sales")
plt.show()
```

```python
ship_date = pd.to_datetime(df["Ship Date"], format="mixed")

monthly_ship_sales = df.copy()
monthly_ship_sales["Ship Date"] = ship_date
monthly_ship_sales = monthly_ship_sales.groupby(
    monthly_ship_sales["Ship Date"].dt.to_period("M")
)["Sales"].sum()

monthly_ship_sales.plot(kind="line", figsize=(10,5))

plt.title("Monthly Sales by Ship Date")
plt.xlabel("Ship Month")
plt.ylabel("Total Sales")
plt.show()
```
